```python
print("="*50)
print("  MCP NEURON: AND GATE")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# AND Gate
def and_gate(x1, x2):
    w1 = 1
    w2 = 1
    T = 1.5
    yin = (x1 * w1) + (x2 * w2)
    return 1 if yin >= T else 0

# NOT Gate
def not_gate(x):
    w1 = -1
    T = -0.5
    yin = x * w1
    return 1 if yin >= T else 0

def and_not(x1, x2):
    """
    Returns 1 if x1=1 and x2=0, else 0
    Equivalent to: x1 AND NOT x2
    """
    not_x2 = not_gate(x2)
    return and_gate(x1, not_x2)

# Test all 4 cases
print("\n  x1   x2  |  y")
print("-"*20)

# Test case 1: 0 0
y = and_not(0, 0)
print(f"  0    0   |  {y}")

# Test case 2: 0 1
y = and_not(0, 1)
print(f"  0    1   |  {y}")

# Test case 3: 1 0
y = and_not(1, 0)
print(f"  1    0   |  {y}")

# Test case 4: 1 1
y = and_not(1, 1)
print(f"  1    1   |  {y}")

print("\n" + "="*50)
print("✅ NOR Gate Implemented Successfully!")
print("="*50)

```
