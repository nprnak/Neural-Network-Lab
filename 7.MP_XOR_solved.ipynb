```python
# ============================================================
# McCULLOCH-PITTS NEURON - XOR GATE (Multi-Layer)
# XOR = AND(NAND(x1,x2), OR(x1,x2))
# Program by: Pranaya Nakarmi
# Roll No: 1
# ============================================================

print("="*50)
print("  MCP NEURON: XOR GATE (Multi-Layer)")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# Layer 1: NAND Gate (Hidden Neuron 1)
def nand_gate(x1, x2):
    w1 = -1
    w2 = -1
    T = -1.5
    yin = (x1 * w1) + (x2 * w2)
    return 1 if yin >= T else 0

# Layer 1: OR Gate (Hidden Neuron 2)
def or_gate(x1, x2):
    w1 = 1
    w2 = 1
    T = 0.5
    yin = (x1 * w1) + (x2 * w2)
    return 1 if yin >= T else 0

# Layer 2: AND Gate (Output Neuron)
def and_gate(x1, x2):
    w1 = 1
    w2 = 1
    T = 1.5
    yin = (x1 * w1) + (x2 * w2)
    return 1 if yin >= T else 0

# XOR Function using Multi-Layer MCP
def xor_gate(x1, x2):
    # Step 1: Calculate hidden layer outputs
    nand_out = nand_gate(x1, x2)    # NAND(x1, x2)
    or_out = or_gate(x1, x2)        # OR(x1, x2)
    
    # Step 2: Calculate output using AND gate
    return and_gate(nand_out, or_out)  # AND(NAND, OR)

print("\n🔍 XOR Truth Table:")
print("   x1   x2  |  NAND  OR  |  XOR")
print("   " + "-"*35)

inputs = [(0,0), (0,1), (1,0), (1,1)]
for x1, x2 in inputs:
    nand_out = nand_gate(x1, x2)
    or_out = or_gate(x1, x2)
    xor_out = xor_gate(x1, x2)
    
    print(f"   {x1}    {x2}   |   {nand_out}     {or_out}   |   {xor_out}")

print("\n" + "="*50)
print("✅ XOR Gate Implemented Successfully!")
print("   Using: XOR = AND(NAND(x1,x2), OR(x1,x2))")
print("="*50)

# Additional Validation
print("\n📊 Validation with different weight combinations:")
print("   XOR Formula: XOR = (x1 NAND x2) AND (x1 OR x2)")
print("   This works because NAND gives 1 except when both are 1,")
print("   and OR gives 1 when at least one is 1.")
print("   AND of these two gives 1 only when both are 1,")
print("   which correctly implements XOR!")
```

```python
# ============================================================
# McCULLOCH-PITTS NEURON - XOR GATE
# Using: AND_NOT and NOT_AND Functions
# XOR = OR(AND_NOT(x1, x2), NOT_AND(x1, x2))
# Program by: Pranaya Nakarmi
# Roll No: 1
# ============================================================

print("="*50)
print("  MCP NEURON: XOR GATE")
print("  Using AND_NOT and NOT_AND Functions")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# ============================================================
# BASIC GATE FUNCTIONS
# ============================================================

# NOT Gate
def not_gate(x):
    w1 = -1
    T = -0.5
    yin = x * w1
    return 1 if yin >= T else 0

# AND Gate
def and_gate(x1, x2):
    w1 = 1
    w2 = 1
    T = 1.5
    yin = (x1 * w1) + (x2 * w2)
    return 1 if yin >= T else 0

# OR Gate
def or_gate(x1, x2):
    w1 = 1
    w2 = 1
    T = 0.5
    yin = (x1 * w1) + (x2 * w2)
    return 1 if yin >= T else 0

# ============================================================
# CUSTOM FUNCTIONS FOR XOR
# ============================================================

# AND_NOT Function: x1 AND NOT x2
def and_not(x1, x2):
    """
    Returns 1 if x1=1 and x2=0, else 0
    Equivalent to: x1 AND NOT x2
    """
    not_x2 = not_gate(x2)
    return and_gate(x1, not_x2)

# NOT_AND Function: NOT x1 AND x2
def not_and(x1, x2):
    """
    Returns 1 if x1=0 and x2=1, else 0
    Equivalent to: NOT x1 AND x2
    """
    not_x1 = not_gate(x1)
    return and_gate(not_x1, x2)

# ============================================================
# XOR GATE IMPLEMENTATION
# XOR = OR(AND_NOT(x1, x2), NOT_AND(x1, x2))
# ============================================================

def xor_gate(x1, x2):
    # Step 1: Calculate AND_NOT and NOT_AND
    term1 = and_not(x1, x2)    # x1 AND NOT x2
    term2 = not_and(x1, x2)    # NOT x1 AND x2
    
    # Step 2: OR the results
    result = or_gate(term1, term2)
    
    return result

# ============================================================
# TEST XOR GATE
# ============================================================

print("\n🔍 XOR Truth Table:")
print("   x1   x2  |  AND_NOT  NOT_AND  |  XOR")
print("   " + "-"*35)

inputs = [(0,0), (0,1), (1,0), (1,1)]
target = [0, 1, 1, 0]

for x1, x2 in inputs:
    term1 = and_not(x1, x2)
    term2 = not_and(x1, x2)
    result = xor_gate(x1, x2)
    
    print(f"   {x1}    {x2}   |    {term1}       {term2}     |    {result}")

# Verification
print("\n" + "="*50)
print("📊 Verification:")
results = [xor_gate(x1, x2) for x1, x2 in inputs]

if results == target:
    print("✅ XOR Gate Implemented Successfully!")
    print("   XOR = OR(AND_NOT(x1, x2), NOT_AND(x1, x2))")
    print("   Where:")
    print("   AND_NOT(x1, x2) = x1 AND NOT x2")
    print("   NOT_AND(x1, x2) = NOT x1 AND x2")
else:
    print("❌ XOR Gate NOT Implemented Correctly!")
    print(f"   Expected: {target}")
    print(f"   Got:      {results}")

print("="*50)

# ============================================================
# DETAILED BREAKDOWN FOR EACH CASE
# ============================================================

print("\n📝 Detailed Breakdown for Each Case:")

print("\n╔═══════════════════════════════════════════════════════╗")
print("║  Case 1: x1=0, x2=0                                 ║")
print("╚═══════════════════════════════════════════════════════╝")
print("  AND_NOT(0, 0) = 0 AND NOT 0")
print("               = 0 AND 1")
print("               = 0")
print("  NOT_AND(0, 0) = NOT 0 AND 0")
print("                = 1 AND 0")
print("                = 0")
print("  XOR = OR(0, 0) = 0 ✓")

print("\n╔═══════════════════════════════════════════════════════╗")
print("║  Case 2: x1=0, x2=1                                 ║")
print("╚═══════════════════════════════════════════════════════╝")
print("  AND_NOT(0, 1) = 0 AND NOT 1")
print("               = 0 AND 0")
print("               = 0")
print("  NOT_AND(0, 1) = NOT 0 AND 1")
print("                = 1 AND 1")
print("                = 1")
print("  XOR = OR(0, 1) = 1 ✓")

print("\n╔═══════════════════════════════════════════════════════╗")
print("║  Case 3: x1=1, x2=0                                 ║")
print("╚═══════════════════════════════════════════════════════╝")
print("  AND_NOT(1, 0) = 1 AND NOT 0")
print("               = 1 AND 1")
print("               = 1")
print("  NOT_AND(1, 0) = NOT 1 AND 0")
print("                = 0 AND 0")
print("                = 0")
print("  XOR = OR(1, 0) = 1 ✓")

print("\n╔═══════════════════════════════════════════════════════╗")
print("║  Case 4: x1=1, x2=1                                 ║")
print("╚═══════════════════════════════════════════════════════╝")
print("  AND_NOT(1, 1) = 1 AND NOT 1")
print("               = 1 AND 0")
print("               = 0")
print("  NOT_AND(1, 1) = NOT 1 AND 1")
print("                = 0 AND 1")
print("                = 0")
print("  XOR = OR(0, 0) = 0 ✓")

print("\n" + "="*50)
print("✅ XOR Gate Implemented Successfully!")
print("   Using: AND_NOT and NOT_AND Functions")
print("="*50)
        
```
