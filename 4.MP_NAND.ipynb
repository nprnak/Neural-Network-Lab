```python
# ============================================================
# McCULLOCH-PITTS NEURON - NAND GATE (Fixed Threshold)
# Program by: Pranaya Nakarmi
# Roll No: 1
# ============================================================

print("="*50)
print("  MCP NEURON: NAND GATE")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)
# MCP Neuron Function for NAND Gate
def mcp_nand(x1, x2):
    w1 = -1          # Negative weights
    w2 = -1
    T = -1.5         # Negative threshold
    yin = (x1 * w1) + (x2 * w2)
    if yin >= T:
        return 1
    else:
        return 0

# Test all 4 cases
print("\n  x1   x2  |  y")
print("-"*20)

# Test case 1: 0 0
y = mcp_nand(0, 0)
print(f"  0    0   |  {y}")

# Test case 2: 0 1
y = mcp_nand(0, 1)
print(f"  0    1   |  {y}")

# Test case 3: 1 0
y = mcp_nand(1, 0)
print(f"  1    0   |  {y}")

# Test case 4: 1 1
y = mcp_nand(1, 1)
print(f"  1    1   |  {y}")

print("\n" + "="*50)
print("✅ NAND Gate Implemented Successfully!")
print("="*50)
        
```

```python
# ============================================================
# McCULLOCH-PITTS NEURON - NAND GATE (Dynamic Threshold)
# Program by: Pranaya Nakarmi
# Roll No: 1
# ============================================================

print("="*50)
print("  MCP NEURON: NAND GATE (Dynamic Threshold)")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# MCP Neuron Function - T is now dynamic
def mcp_nand(x1, x2, T):
    w1 = -1          # Negative weights
    w2 = -1
    yin = (x1 * w1) + (x2 * w2)
    if yin >= T:
        return 1
    else:
        return 0

# Let user set the threshold
T = float(input("Enter threshold value (T): "))

# Expected NAND gate outputs for (0,0),(0,1),(1,0),(1,1)
target = [1, 1, 1, 0]

inputs = [(0,0), (0,1), (1,0), (1,1)]
outputs = []

print(f"\nUsing Threshold T = {T}")
print("\n  x1   x2  |  y")
print("-"*20)

for x1, x2 in inputs:
    y = mcp_nand(x1, x2, T)
    outputs.append(y)
    print(f"  {x1}    {x2}   |  {y}")

print("\n" + "="*50)
if outputs == target:
    print("✅ NAND Gate Implemented Successfully!")
else:
    print("❌ NAND Gate NOT Implemented — output doesn't match target")
    print(f"   Expected: {target}")
    print(f"   Got:      {outputs}")
print("="*50)
        
```

```python
# ============================================================
# McCULLOCH-PITTS NEURON - NAND GATE (Dynamic Weights & Threshold)
# Program by: Pranaya Nakarmi
# Roll No: 1
# ============================================================

print("="*50)
print("  MCP NEURON: NAND GATE (Dynamic Weights & Threshold)")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# MCP Neuron Function - w1, w2, T are all dynamic
def mcp_nand(x1, x2, w1, w2, T):
    yin = (x1 * w1) + (x2 * w2)
    if yin >= T:
        return 1
    else:
        return 0

# Expected NAND gate outputs for (0,0),(0,1),(1,0),(1,1)
target = [1, 1, 1, 0]
inputs = [(0,0), (0,1), (1,0), (1,1)]

attempt = 1
success = False

while not success:
    print(f"\n--- Attempt {attempt} ---")
    w1 = float(input("Enter weight w1: "))
    w2 = float(input("Enter weight w2: "))
    T = float(input("Enter threshold value (T): "))

    outputs = []
    print(f"\nUsing w1={w1}, w2={w2}, T={T}")
    print("\n  x1   x2  |  y")
    print("-"*20)

    for x1, x2 in inputs:
        y = mcp_nand(x1, x2, w1, w2, T)
        outputs.append(y)
        print(f"  {x1}    {x2}   |  {y}")

    print("\n" + "="*50)
    if outputs == target:
        success = True
        print("✅ NAND Gate Implemented Successfully!")
        print(f"   Final values -> w1={w1}, w2={w2}, T={T}")
    else:
        print("❌ NAND Gate NOT Implemented — output doesn't match target")
        print(f"   Expected: {target}")
        print(f"   Got:      {outputs}")
        print("   Please try different values of w1, w2, T.")
    print("="*50)

    attempt += 1
```
