```python
print("="*50)
print("  MCP NEURON: AND GATE")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# MCP Neuron Function
def mcp_and(x1, x2):
    w1 = 1
    w2 = 1
    T = 1.5
    yin = (x1 * w1) + (x2 * w2)
    if yin >= T:
        return 1
    else:
        return 0

# Test all 4 cases
print("\n  x1   x2  |  y")
print("-"*20)

# Test case 1: 0 0
y = mcp_and(0, 0)
print(f"  0    0   |  {y}")

# Test case 2: 0 1
y = mcp_and(0, 1)
print(f"  0    1   |  {y}")

# Test case 3: 1 0
y = mcp_and(1, 0)
print(f"  1    0   |  {y}")

# Test case 4: 1 1
y = mcp_and(1, 1)
print(f"  1    1   |  {y}")

print("\n" + "="*50)
print("✅ AND Gate Implemented Successfully!")
print("="*50)
        
```

```python
print("="*50)
print("  MCP NEURON: AND GATE")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

# MCP Neuron Function - T is now dynamic
def mcp_and(x1, x2, T):
    w1 = 1
    w2 = 1
    yin = (x1 * w1) + (x2 * w2)
    if yin >= T:
        return 1
    else:
        return 0

# Let user set the threshold
T = float(input("Enter threshold value (T): "))

# Expected AND gate outputs for (0,0),(0,1),(1,0),(1,1)
target = [0, 0, 0, 1]

inputs = [(0,0), (0,1), (1,0), (1,1)]
outputs = []

print(f"\nUsing Threshold T = {T}")
print("\n  x1   x2  |  y")
print("-"*20)

for x1, x2 in inputs:
    y = mcp_and(x1, x2, T)
    outputs.append(y)
    print(f"  {x1}    {x2}   |  {y}")

print("\n" + "="*50)
if outputs == target:
    print("✅ AND Gate Implemented Successfully!")
else:
    print("❌ AND Gate NOT Implemented — output doesn't match target")
    print(f"   Expected: {target}")
    print(f"   Got:      {outputs}")
print("="*50)
```
