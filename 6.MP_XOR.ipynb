```python
# ============================================================
# McCULLOCH-PITTS NEURON - XOR GATE (Attempt)
# Program by: Pranaya Nakarmi
# Roll No: 1
# ============================================================

print("="*50)
print("  MCP NEURON: XOR GATE (Attempt)")
print("  Program by: Pranaya Nakarmi")
print("  Roll No: 1")
print("="*50)

print("\n🔍 Trying to find weights and threshold for XOR...")
print("   XOR Truth Table:")
print("   (0,0) → 0")
print("   (0,1) → 1")
print("   (1,0) → 1")
print("   (1,1) → 0")

# Function to test XOR with given weights and threshold
def test_xor(w1, w2, T):
    print(f"\n📊 Testing with w1={w1}, w2={w2}, T={T}")
    print("   x1   x2  |  yin  |  y")
    print("   " + "-"*25)
    
    inputs = [(0,0), (0,1), (1,0), (1,1)]
    target = [0, 1, 1, 0]
    outputs = []
    
    for x1, x2 in inputs:
        yin = (x1 * w1) + (x2 * w2)
        y = 1 if yin >= T else 0
        outputs.append(y)
        print(f"   {x1}    {x2}   |  {yin:3}  |  {y}")
    
    print("   " + "-"*25)
    print(f"   Output:  {outputs}")
    print(f"   Target:  {target}")
    
    if outputs == target:
        print("   ✅ XOR Implemented Successfully!")
        return True
    else:
        print("   ❌ XOR NOT Implemented!")
        return False

# Try different combinations
print("\n" + "="*50)
print("🔬 TEST 1: AND-like weights (1, 1)")
test_xor(1, 1, 1.5)

print("\n" + "="*50)
print("🔬 TEST 2: OR-like weights (1, 1)")
test_xor(1, 1, 0.5)

print("\n" + "="*50)
print("🔬 TEST 3: NAND-like weights (-1, -1)")
test_xor(-1, -1, -0.5)

print("\n" + "="*50)
print("🔬 TEST 4: Trying custom weights (1, -1)")
test_xor(1, -1, 0)

print("\n" + "="*50)
print("🔬 TEST 5: Trying custom weights (-1, 1)")
test_xor(-1, 1, 0)

print("\n" + "="*50)
print("🔬 TEST 6: Trying custom weights (2, 2)")
test_xor(2, 2, 3)

print("\n" + "="*50)
print("🔬 TEST 7: Trying custom weights (-2, -2)")
test_xor(-2, -2, -1)

print("\n" + "="*50)
print("📌 CONCLUSION: XOR gate CANNOT be implemented")
print("   with a single MCP neuron!")
print("="*50)
        
```
